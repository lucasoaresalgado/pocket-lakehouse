# NASA POWER — daily climate per municipality

Regional endpoint (blocks of up to 10° x 10°), nearest grid point per municipality. Loading some parameters keeps the others (partial update). Table `nasa_power_diario`, partitioned by year.

```python
from pprint import pprint
from pipelines import power
from lakehouse import core as C
```

## Parameters

```python
params = dict(mode="incremental", reference_date="", start_date="", end_date="", lookback_days=75, max_days=366,
              parametros=["PRECTOTCORR"], tile_deg=10, reset_cursor=False, target_db="estudo", ref_db="")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("power")
resultado = power.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
