# INPE/CPTEC MERGE — daily rain per municipality

Daily GRIB2 (0.1°) decoded in numpy, nearest pixel + 3x3 mean per IBGE municipality. Table `cptec_merge_diario`, partitioned by year.

```python
from pprint import pprint
from pipelines import merge
from lakehouse import core as C
```

## Parameters

```python
params = dict(mode="incremental", reference_date="", start_date="", end_date="", lookback_days=7, max_days=120,
              concurrency=2, force=False, reset_cursor=False, target_db="estudo", ref_db="")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("merge")
resultado = merge.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
