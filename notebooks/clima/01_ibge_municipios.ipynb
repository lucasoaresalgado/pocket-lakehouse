# IBGE — municipalities, states and aliases (centroids)

Reference table for the climate pipelines (they sample each municipality at its centroid). Snapshot of states and municipalities; aliases are upserted.

```python
from pprint import pprint
from pipelines import ibge
from lakehouse import core as C
```

## Parameters

```python
params = dict(target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("ibge")
resultado = ibge.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
