# Receita Federal — CNPJ open data

Snapshot per table (`cnpj_*`). The full release is ~7 GB zipped: use `ufs` (establishments of those states; Empresas/Socios/Simples follow them) and/or `parts` (0-9). Without a filter the size guard refuses a load that would not fit.

```python
from pprint import pprint
from pipelines import cnpj
from lakehouse import core as C
```

## Parameters

```python
params = dict(reference_date="", tables=None, parts=None, ufs=["ES"], ignorar_cota=False, manter_cpf_aberto=False, target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("cnpj")
resultado = cnpj.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
