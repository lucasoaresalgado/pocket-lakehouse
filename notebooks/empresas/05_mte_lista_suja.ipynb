# MTE — Cadastro de Empregadores (Lista Suja)

Spreadsheet parsed with openpyxl, upserted by `chave_registro`. Open CPFs of individuals are stored masked unless `manter_cpf_aberto=True`.

```python
from pprint import pprint
from pipelines import mte
from lakehouse import core as C
```

## Parameters

```python
params = dict(full_refresh=False, manter_cpf_aberto=False, target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("mte")
resultado = mte.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
