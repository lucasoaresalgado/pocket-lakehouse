# CVM — financial statements of listed companies (DFP/ITR)

One table per statement (`cvm_{dfp|itr}_{statement}`), snapshot of the years of the run.

```python
from pprint import pprint
from pipelines import cvm
from lakehouse import core as C
```

## Parameters

```python
params = dict(years=[2025], doc_types=["DFP", "ITR"], target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("cvm")
resultado = cvm.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
