# Compras.gov.br / PNCP — contratações, itens e resultados (Lei 14.133)

`mode="api"` = incremental by date (the mirror lags ~4 weeks in 10/2026); `mode="csv"` = monthly zips (`csv_months` / `csv_years`).

```python
from pprint import pprint
from pipelines import pncp
from lakehouse import core as C
```

## Parameters

```python
params = dict(mode="api", datasets=None, lookback_days=3, date_start="", date_end="", csv_months=None, csv_years=None,
              full_refresh=False, manter_cpf_aberto=False, target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("pncp")
resultado = pncp.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
