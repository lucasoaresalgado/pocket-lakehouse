# CGU — sanction registries (CEIS, CNEP, CEPIM, Leniência, CEAF)

Daily snapshots upserted by PK; `data_referencia` = last snapshot in which a row was seen. The dry run downloads and validates the headers (no Spark).

```python
from pprint import pprint
from pipelines import cgu
from lakehouse import core as C
```

## Parameters

```python
params = dict(cadastros=None, offset_days=2, reference_date="", full_refresh=False, manter_cpf_aberto=False, target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("cgu")
resultado = cgu.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
