# PGFN — Lista de Devedores (dívida ativa da União)

Three quarterly zips into `pgfn_devedores` (+ `tipo_divida`), snapshot of the quarter. `ufs` filters the debtors' states.

```python
from pprint import pprint
from pipelines import pgfn
from lakehouse import core as C
```

## Parameters

```python
params = dict(tipos=None, quarter="", ufs=[], ignorar_cota=False, target_db="estudo")
dry_run = True   # False = baixa a fonte e grava Parquet no MinIO (lake/<target_db>/)
```

## Run

```python
spark = None if dry_run else C.get_spark("pgfn")
resultado = pgfn.main(spark, **params, dry_run=dry_run)
pprint(resultado, sort_dicts=False, width=140)
```
