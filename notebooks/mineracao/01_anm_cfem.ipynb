# ANM CFEM — collection


```python
from pprint import pprint
from pipelines import common as C, cfem
```

## Parameters

```python
params = dict(mode="incremental", reference_date="", start_date="", end_date="", lookback_meses=6, incluir_p3=True,
              force=False, target_db="bronze", cfg_db="")
dry_run = True   # False = download the source and write Parquet to MinIO (lake/<target_db>/)
profile = False  # True = read and profile without writing (needs dry_run=False)
```

## Run

```python
spark = None if dry_run else C.get_spark("cfem")
resultado = cfem.main(spark, **params, dry_run=dry_run, profile=profile)
pprint(resultado, sort_dicts=False, width=140)
```
