# ANM Mining Registry (SCM)

```python
from pprint import pprint
from pipelines import common as C, scm
```

## Parameters

```python
params = dict(mode="incremental", incluir_p3=True, force=False, target_db="bronze", cfg_db="")
dry_run = True   # False = download the source and write Parquet to MinIO (lake/<target_db>/)
profile = False  # True = read and profile without writing (needs dry_run=False)
```

## Run

```python
spark = None if dry_run else C.get_spark("scm")
resultado = scm.main(spark, **params, dry_run=dry_run, profile=profile)
pprint(resultado, sort_dicts=False, width=140)
```

> **Hardware:** the SCM ZIP is ~1 GB uncompressed and the events table is the largest. It works with 2 executors of 1.5 GB, but takes minutes and uses ~1.2 GB of disk in `/data/raw/anm_scm` (removed at the end; `KEEP_RAW=1` keeps it). Run with free memory and nothing else heavy alongside.
